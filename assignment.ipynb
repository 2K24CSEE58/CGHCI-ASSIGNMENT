```python
Problem 1: Thresholding
Step 1: Inputs and output

Input mein grayscale image aur threshold value hoti hai. Output mein black and white image milti hai, jisme values sirf 0 ya 255 hoti hain.

Step 2: Rule

Agar pixel ki value threshold ke equal ya us se zyada ho, to output 255 hoga. Warna output 0 hoga.

Step 3: Small example

Threshold 128 hai: 20   128   200
100  150   250  Isliye result hoga:

0   255   255
0   255   255
Step 4: Python
import numpy as np

def threshold_image(image: np.ndarray, threshold: int) -> np.ndarray:
    Convert a grayscale image to black and white.
    return np.where(image >= threshold, 255, 0).astype(np.uint8)


problem_1_input = np.array([[20, 128, 200],
                            [100, 150, 250]], dtype=np.uint8)

problem_1_expected = np.array([[0, 255, 255],
                               [0, 255, 255]], dtype=np.uint8)

np.testing.assert_array_equal(
    threshold_image(problem_1_input, 128),
    problem_1_expected
)

print("Problem 1 passed")

Output:

Problem 1 passed
```

```python
Problem 2: Image Memory
Step 1: Inputs and output

Input mein image ki width, height aur bits per pixel hotay hain. Output image ki memory bytes mein hoti hai.

Step 2: Rule

Memory calculate karne ke liye:

bytes = width × height × bpp / 8
Step 3: Small example

Agar width 1920, height 1080 aur 24 bpp ho:

1920 × 1080 × 24 / 8
= 6,220,800 bytes
Step 4: Python
def display_memory_bytes(width: int, height: int, bpp: int) -> int:
    """Return image memory in bytes."""
    return width * height * bpp // 8


expected_bytes = 1920 * 1080 * 24 // 8

assert display_memory_bytes(1920, 1080, 24) == expected_bytes

print("Problem 2 passed")

Output:

Problem 2 passed
```

```python
Problem 3: Average 9 Pixels
Step 1: Inputs and output

Input mein 3 × 3 image block hota hai jisme total 9 pixel values hoti hain. Output un 9 pixels ka rounded average hota hai.

Step 2: Rule

Sabhi 9 values ko add karke 9 se divide karte hain.

average = sum of all pixels / 9
Step 3: Small example
10  20  10
30  50  30
10  20  10

Sum:

10 + 20 + 10 + 30 + 50 + 30 + 10 + 20 + 10 = 190

Average:

190 / 9 = 21.11

Rounded answer 21 hai.

Step 4: Python
def mean_filter_3x3(neighborhood: np.ndarray) -> int:
    """Return the rounded average of 9 pixels."""
    return int(round(np.sum(neighborhood) / 9))


problem_3_input = np.array([[10, 20, 10],
  [30, 50, 30],
 [10, 20, 10]], dtype=np.uint8)

assert mean_filter_3x3(problem_3_input) == 21

print("Problem 3 passed")

Output:

Problem 3 passed
```

```python
Problem 4: Contrast Stretching
Step 1: Inputs and output

Input mein image values aur input range hoti hai. Output mein image values ko ek new range mein convert kiya jata hai.

Step 2: Rule

Formula hai:

output = ((input - in_min) / (in_max - in_min))
         × (out_max - out_min) + out_min

Agar result output range se bahar ho to usay range ke andar clip kar dete hain.

Step 3: Small example

Input values:

50, 100, 150

Input range 50–150 aur output range 0–255 hai.

Isliye:

50  → 0
100 → 127.5
150 → 255
Step 4: Python
def contrast_stretch(
    image: np.ndarray,
    in_min: float,
    in_max: float,
    out_min: float = 0.0,
    out_max: float = 255.0,
) -> np.ndarray:
    """Change image values from one range to another."""
    
    stretched = ((image - in_min) / (in_max - in_min)) * \
                (out_max - out_min) + out_min
    
    return np.clip(stretched, out_min, out_max)


problem_4_input = np.array([50, 100, 150], dtype=np.float32)

expected = np.array([0.0, 127.5, 255.0], dtype=np.float32)

np.testing.assert_allclose(
    contrast_stretch(problem_4_input, 50, 150),
    expected
)

print("Problem 4 passed")

Output:

Problem 4 passed
```

```python
Problem 5: Find an Edge with Sobel
Step 1: Inputs and output

Input mein 3 × 3 grayscale image block hota hai. Output mein gx, gy aur edge ki strength milti hai.

Step 2: Rule

Sobel ke liye do kernels use hote hain, Gx aur Gy. Dono kernels ko image block ke saath multiply karke values add karte hain.

Phir edge strength:

strength = sqrt(gx² + gy²)
Step 3: Small example

Given block:

20  20  200
20  20  200
20  20  200

Is example mein:

gx = 720
gy = 0

Isliye:

strength = sqrt(720² + 0²)
         = 720
Step 4: Python
def sobel_response(block: np.ndarray) -> tuple[float, float, float]:
    """Return gx, gy, and edge strength."""
    
    gx_kernel = np.array([
        [-1, 0, 1],
        [-2, 0, 2],
        [-1, 0, 1]
    ], dtype=np.float32)

    gy_kernel = np.array([
        [-1, -2, -1],
        [0, 0, 0],
        [1, 2, 1]
    ], dtype=np.float32)

    gx = float(np.sum(block * gx_kernel))
    gy = float(np.sum(block * gy_kernel))

    strength = float(np.sqrt(gx * gx + gy * gy))

    return gx, gy, strength


problem_5_input = np.array([
    [20, 20, 200],
    [20, 20, 200],
    [20, 20, 200]
], dtype=np.float32)

gx, gy, strength = sobel_response(problem_5_input)

assert gx == 720.0 and gy == 0.0

print("Problem 5 passed")

Output:

Problem 5 passed
```
